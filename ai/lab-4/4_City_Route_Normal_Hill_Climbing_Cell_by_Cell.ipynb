# 4-City Route Optimization using Normal Hill Climbing

A delivery person starts at **A**, visits **B, C, and D** exactly once, and returns to **A**. The objective is to minimize total travel cost.

The program is divided into small cells so that each step can be run and explained separately.

## Example: How to enter the problem

Use the following cost matrix:

| From / To | A | B | C | D |
|---|---:|---:|---:|---:|
| **A** | 0 | 12 | 18 | 10 |
| **B** | 12 | 0 | 20 | 15 |
| **C** | 18 | 20 | 0 | 25 |
| **D** | 10 | 15 | 25 | 0 |

The starting city is **A**.

### Example route

Enter:

```text
B C D
```

This means:

```text
A → B → C → D → A
```

### Example cost calculation

For `A → B → C → D → A`:

```text
A → B = 12
B → C = 20
C → D = 25
D → A = 10

Total = 12 + 20 + 25 + 10 = 67
```

### Example neighbour

Starting route:

```text
A → B → C → D → A
```

Swap B and C:

```text
A → C → B → D → A
```

Its cost is:

```text
A → C = 18
C → B = 20
B → D = 15
D → A = 10

Total = 63
```

Since `63 < 67`, this neighbour is better, so Hill Climbing moves to it.

The program will continue checking neighbours until it cannot find a better route.


````python
# Cell 1: Cost matrix

def create_costs():
    return {
        'A': {'A': 0, 'B': 12, 'C': 18, 'D': 10},
        'B': {'A': 12, 'B': 0, 'C': 20, 'D': 15},
        'C': {'A': 18, 'B': 20, 'C': 0, 'D': 25},
        'D': {'A': 10, 'B': 15, 'C': 25, 'D': 0}
    }

costs = create_costs()
costs
````

## Step 2: Take the starting route

The next cell asks the user to enter the initial route.

````python
# Cell 2: Function to take the starting route

def take_route_input():
    print("Cities available: B, C, D")
    print("Starting city A is fixed.")
    print("Example input: B C D")

    while True:
        route = input("Enter starting route: ").upper().split()

        if len(route) == 3 and sorted(route) == ['B', 'C', 'D']:
            return route

        print("Invalid input. Enter B, C and D exactly once.")

start_route = take_route_input()

print("You entered:", "A → " + " → ".join(start_route) + " → A")
````

## Step 3: Calculate the cost

The cost function adds the cost of every journey, including the return to A.

````python
# Cell 3: Calculate route cost

def route_cost(route):
    full_route = ['A'] + route + ['A']
    total = 0

    for i in range(len(full_route) - 1):
        total += costs[full_route[i]][full_route[i + 1]]

    return total

print("Starting route cost:", route_cost(start_route))
````

## Step 4: Generate neighbours

Each neighbour is made by swapping two cities.

````python
# Cell 4: Generate neighbouring routes

def get_neighbors(route):
    neighbors = []

    for i in range(len(route)):
        for j in range(i + 1, len(route)):
            new_route = route.copy()
            new_route[i], new_route[j] = new_route[j], new_route[i]
            neighbors.append(new_route)

    return neighbors

neighbors = get_neighbors(start_route)

print("Neighbours:")
for route in neighbors:
    print("A → " + " → ".join(route) + " → A",
          "| Cost:", route_cost(route))
````

## Step 5: Apply Hill Climbing

````python
# Cell 5: Normal Hill Climbing

def hill_climbing(start_route):
    current = start_route.copy()

    while True:
        neighbors = get_neighbors(current)

        best_neighbor = min(neighbors, key=route_cost)

        if route_cost(best_neighbor) >= route_cost(current):
            break

        current = best_neighbor

    return current

solution = hill_climbing(start_route)

print("Final solution:")
print("A → " + " → ".join(solution) + " → A")
print("Minimum cost:", route_cost(solution))
````

## Reading the solution

The program follows these steps:

```text
Starting route
      ↓
Calculate cost
      ↓
Generate neighbours
      ↓
Find best neighbour
      ↓
Is it better?
   ↙        ↘
 Yes         No
 ↓           ↓
Move       Stop
 ↓
Repeat
```

The final route is the route where **no neighbouring route has a lower cost**.
