# 4-City Route Optimization using Random-Restart Hill Climbing

This notebook solves the same problem, but instead of using only one starting route, it performs **multiple Hill Climbing runs from random starting routes**.

The user manually enters the **number of random restarts**.


## Example: How to enter the problem

Use the following cost matrix:

| From / To | A | B | C | D |
|---|---:|---:|---:|---:|
| **A** | 0 | 12 | 18 | 10 |
| **B** | 12 | 0 | 20 | 15 |
| **C** | 18 | 20 | 0 | 25 |
| **D** | 10 | 15 | 25 | 0 |

The starting city is **A**.

### Example route

Enter:

```text
B C D
```

This means:

```text
A → B → C → D → A
```

### Example cost calculation

For `A → B → C → D → A`:

```text
A → B = 12
B → C = 20
C → D = 25
D → A = 10

Total = 12 + 20 + 25 + 10 = 67
```

### Example neighbour

Starting route:

```text
A → B → C → D → A
```

Swap B and C:

```text
A → C → B → D → A
```

Its cost is:

```text
A → C = 18
C → B = 20
B → D = 15
D → A = 10

Total = 63
```

Since `63 < 67`, this neighbour is better, so Hill Climbing moves to it.

The program will continue checking neighbours until it cannot find a better route.


````python
# Cell 1: Cost matrix

def create_costs():
    return {
        'A': {'A': 0, 'B': 12, 'C': 18, 'D': 10},
        'B': {'A': 12, 'B': 0, 'C': 20, 'D': 15},
        'C': {'A': 18, 'B': 20, 'C': 0, 'D': 25},
        'D': {'A': 10, 'B': 15, 'C': 25, 'D': 0}
    }

costs = create_costs()
costs
````

## Step 2: Enter the number of random restarts

For example, enter `5` to perform five random starts.

````python
# Cell 2: Function to manually enter the number of random restarts

def take_restart_input():
    while True:
        try:
            restarts = int(input("Enter number of random restarts: "))

            if restarts > 0:
                return restarts

            print("Please enter a positive number.")

        except ValueError:
            print("Please enter a whole number.")

restarts = take_restart_input()

print("Number of random restarts:", restarts)
````

## Step 3: Cost and neighbour functions

````python
# Cell 3: Route cost and neighbour functions

def route_cost(route):
    full_route = ['A'] + route + ['A']
    total = 0

    for i in range(len(full_route) - 1):
        total += costs[full_route[i]][full_route[i + 1]]

    return total


def get_neighbors(route):
    neighbors = []

    for i in range(len(route)):
        for j in range(i + 1, len(route)):
            new_route = route.copy()
            new_route[i], new_route[j] = new_route[j], new_route[i]
            neighbors.append(new_route)

    return neighbors
````

## Step 4: Hill Climbing function

````python
# Cell 4: Hill Climbing function

def hill_climbing(start_route):
    current = start_route.copy()

    while True:
        neighbors = get_neighbors(current)
        best_neighbor = min(neighbors, key=route_cost)

        if route_cost(best_neighbor) >= route_cost(current):
            break

        current = best_neighbor

    return current
````

## Step 5: Run the random restarts

````python
# Cell 5: Run Hill Climbing from random starts

import random

best_route = None
best_cost = float('inf')

for i in range(restarts):
    start_route = random.sample(['B', 'C', 'D'], 3)
    solution = hill_climbing(start_route)
    solution_cost = route_cost(solution)

    print(f"Restart {i + 1}")
    print("  Start :", "A → " + " → ".join(start_route) + " → A")
    print("  Local optimum :", "A → " + " → ".join(solution) + " → A")
    print("  Cost :", solution_cost)
    print()

    if solution_cost < best_cost:
        best_cost = solution_cost
        best_route = solution

print("--- Best Overall Solution ---")
print("A → " + " → ".join(best_route) + " → A")
print("Minimum cost:", best_cost)
````

If you enter:


```text
Random Start 1
      ↓
Hill Climbing
      ↓
Local Optimum

Random Start 2
      ↓
Hill Climbing
      ↓
Local Optimum

Random Start 3
      ↓
Hill Climbing
      ↓
Local Optimum

Random Start 4
      ↓
Hill Climbing
      ↓
Local Optimum

Random Start 5
      ↓
Hill Climbing
      ↓
Local Optimum

       ↓
Compare all results
       ↓
Best route
```

The important idea is that **each restart starts from a different random route**, and the final answer is the best local optimum obtained.
