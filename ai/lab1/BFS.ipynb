```python
from collections import deque

# -------------------------------
# Breadth First Search (BFS)
# -------------------------------

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    print("\nBFS Traversal:")

    while queue:
        # Remove the first node from the queue
        node = queue.popleft()

        # Visit the node if it has not been visited
        if node not in visited:
            print(node, end=" ")

            visited.add(node)

            # Add unvisited neighbours to the queue
            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)


# -------------------------------
# Taking Graph Input
# -------------------------------

graph = {}

n = int(input("Enter number of nodes: "))

print("\nEnter the node names:")
for i in range(n):
    node = input(f"Node {i+1}: ")
    graph[node] = []

e = int(input("\nEnter number of edges: "))

print("\nEnter each edge as: source destination")
for i in range(e):
    u, v = input(f"Edge {i+1}: ").split()

    # Undirected graph
    graph[u].append(v)
    graph[v].append(u)


# -------------------------------
# Display Graph
# -------------------------------

print("\nGraph:")
for node in graph:
    print(node, "->", graph[node])


# -------------------------------
# Starting Node
# -------------------------------

start = input("\nEnter starting node for BFS: ")

bfs(graph, start)
```
