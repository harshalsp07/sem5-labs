```python
alpha_beta_terminal_count = 0

# the below function is just for returning available moves it returns list {1,2} if state = 2
def get_legal_moves(state):
    return [move for move in (1,2,3) if state - move >= 0]

def minimax(state,maximizing_player):
    if state == 0:
        return -1 if maximizing_player else 1

    legal_moves  = get_legal_moves(state)

    if maximizing_player:
        best_score = -float("inf")
        for move in legal_moves:
            score = minimax(state-move,False)
            best_score = max(best_score,score)
        return best_score
    else:
        best_score = float("inf")
        for move in legal_moves:
            score = minimax(state-move , True)
            best_score = min(best_score,score)
        return best_score

def alpha_beta(state,maximizing_player,alpha,beta):

    global alpha_beta_terminal_count

    if state == 0:
        alpha_beta_terminal_count += 1
        return -1 if maximizing_player else 1

    legal_moves = get_legal_moves(state)

    if maximizing_player:
        best_score = -float("inf")
        for move in legal_moves:
            score = alpha_beta(state-move,False,alpha,beta)
            best_score=max(best_score,score)
            alpha = max(alpha,best_score)
            if beta <= alpha:
                break
        return best_score
    else:
        best_score = float("inf")
        for move in legal_moves:
            score = alpha_beta(state-move,True,alpha,beta)
            best_score = min(best_score,score)
            beta = min(beta,best_score)
            if beta <= alpha:
                break
        return best_score

def find_best_move(state):

    best_score = -float("inf")
    best_m = None
    legal_moves = get_legal_moves(state)

    for move in legal_moves:
        score = minimax(state-move,False)
        if score > best_score:
            best_score = score
            best_m = move

    return best_score,best_m

def solve(N):
    """Run tests and display formatted outputs."""
    global alpha_beta_terminal_count
    alpha_beta_terminal_count = 0

    optimal_value, best_m = find_best_move(N)
    alpha_beta(N, True, -float("inf"), float("inf"))

    print(f"Countdown N = {N}")
    print(f"Optimal Value: {optimal_value}")
    print(f"Best Move: {best_m}")
    print(f"Alpha-Beta States: {alpha_beta_terminal_count}")
    print("-" * 30)
    return optimal_value, best_m, alpha_beta_terminal_count
    
    
```

```python
# Execute Sample Test Cases
for test_n in [1, 2, 4, 5, 6]:
    solve(test_n)
```
