```python
# ==========================================
# TIC-TAC-TOE AI USING MINIMAX
# ==========================================

def check_winner(board, player):

    winning_positions = [
        (0, 1, 2),  # Top row
        (3, 4, 5),  # Middle row
        (6, 7, 8),  # Bottom row
        (0, 3, 6),  # Left column
        (1, 4, 7),  # Middle column
        (2, 5, 8),  # Right column
        (0, 4, 8),  # Main diagonal
        (2, 4, 6)   # Other diagonal
    ]

    for a, b, c in winning_positions:

        if (board[a] == player and
            board[b] == player and
            board[c] == player):

            return True

    return False


def minimax(board, maximizing_player):

    
    # 1. CHECK TERMINAL STATES
    

    # If AI (O) wins
    if check_winner(board, "O"):
        return 1

    # If Human (X) wins
    if check_winner(board, "X"):
        return -1

    # If board is full, it is a draw
    if " " not in board:
        return 0


    
    # 2. AI'S TURN - MAXIMIZING PLAYER
    

    if maximizing_player:

        # Start with the worst possible score
        best_score = -float("inf")

        # Try all 9 positions
        for i in range(9):

            # Only play in an empty position
            if board[i] == " ":

                # Make a temporary AI move
                board[i] = "O"

                # Ask Minimax what happens after this move
                # False means it is now the human's turn
                score = minimax(board, False)

                # Undo the temporary move
                board[i] = " "

                # Keep the highest score
                best_score = max(best_score, score)

        return best_score


    
    # 3. HUMAN'S TURN - MINIMIZING PLAYER
    

    else:

        # Start with the best possible score
        best_score = float("inf")

        # Try all 9 positions
        for i in range(9):

            # Only play in an empty position
            if board[i] == " ":

                # Make a temporary human move
                board[i] = "X"

                # Ask Minimax what happens after this move
                # True means it is now the AI's turn
                score = minimax(board, True)

                # Undo the temporary move
                board[i] = " "

                # Keep the lowest score
                best_score = min(best_score, score)

        return best_score


def best_move(board):

    # Start with the worst possible score
    best_score = -float("inf")

    # No move selected yet
    move = None

    # Check all 9 positions
    for i in range(9):

        # Only consider empty positions
        if board[i] == " ":

            # Temporarily place AI's O
            board[i] = "O"

            # Find out how good this move is
            score = minimax(board, False)

            # Undo the move
            board[i] = " "

            # If this move is better than previous moves
            if score > best_score:

                # Save its score
                best_score = score

                # Save its position
                move = i

    return move


def print_board(board):

    print()
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print()


def human_move(board):

    while True:

        try:
            position = int(input("Enter your move (1-9): "))

            # Convert 1-9 into 0-8
            position = position - 1

            # Check whether position is valid
            if position < 0 or position > 8:
                print("Please enter a number from 1 to 9.")
                continue

            # Check whether position is empty
            if board[position] != " ":
                print("That position is already occupied.")
                continue

            # Make human move
            board[position] = "X"

            break

        except ValueError:
            print("Please enter a valid number.")


def main():

    # Empty board
    board = [
        " ", " ", " ",
        " ", " ", " ",
        " ", " ", " "
    ]

    print("================================")
    print("       TIC-TAC-TOE AI")
    print("================================")
    print()
    print("You are X")
    print("AI is O")
    print()
    print("Board positions are:")
    print()
    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")
    print()

    # Human goes first
    current_player = "X"

    while True:

        # ==========================================
        # HUMAN TURN
        # ==========================================

        if current_player == "X":

            print_board(board)

            human_move(board)

            # Check if human won
            if check_winner(board, "X"):

                print_board(board)
                print("Congratulations! You win!")
                break

            # Check for draw
            if " " not in board:

                print_board(board)
                print("It's a draw!")
                break

            # Switch to AI
            current_player = "O"


        # ==========================================
        # AI TURN
        # ==========================================

        else:

            print("\nAI is thinking...")

            move = best_move(board)

            # Make AI move
            board[move] = "O"

            print("AI chose position:", move + 1)

            # Check if AI won
            if check_winner(board, "O"):

                print_board(board)
                print("AI wins!")
                break

            # Check for draw
            if " " not in board:

                print_board(board)
                print("It's a draw!")
                break

            # Switch back to human
            current_player = "X"


# ==========================================
# START THE PROGRAM
# ==========================================

if __name__ == "__main__":
    main()
```

```python

# TIC-TAC-TOE AI USING ALPHA-BETA PRUNING




# CHECK WINNER


def check_winner(board, player):

    winning_positions = [
        (0, 1, 2),  # Top row
        (3, 4, 5),  # Middle row
        (6, 7, 8),  # Bottom row
        (0, 3, 6),  # Left column
        (1, 4, 7),  # Middle column
        (2, 5, 8),  # Right column
        (0, 4, 8),  # Main diagonal
        (2, 4, 6)   # Other diagonal
    ]

    for a, b, c in winning_positions:

        if (board[a] == player and
            board[b] == player and
            board[c] == player):

            return True

    return False


# ALPHA-BETA PRUNING


def minimax(board, maximizing_player, alpha, beta):

    
    # CHECK TERMINAL STATES
    

    # AI wins
    if check_winner(board, "O"):
        return 1

    # Human wins
    if check_winner(board, "X"):
        return -1

    # Draw
    if " " not in board:
        return 0


    
    # AI'S TURN - MAXIMIZING PLAYER
    

    if maximizing_player:

        # Start with the worst possible score
        best_score = -float("inf")

        # Try every position
        for i in range(9):

            # Only use empty positions
            if board[i] == " ":

                # Make AI move
                board[i] = "O"

                # Recursively calculate score
                score = minimax(
                    board,
                    False,
                    alpha,
                    beta
                )

                # Undo move
                board[i] = " "

                # Keep highest score
                best_score = max(best_score, score)

                # Update alpha
                alpha = max(alpha, best_score)

                # ==================================
                # ALPHA-BETA PRUNING
                # ==================================

                if beta <= alpha:
                    break

        return best_score


    
    # HUMAN'S TURN - MINIMIZING PLAYER
    

    else:

        # Start with the highest possible score
        best_score = float("inf")

        # Try every position
        for i in range(9):

            # Only use empty positions
            if board[i] == " ":

                # Make human move
                board[i] = "X"

                # Recursively calculate score
                score = minimax(
                    board,
                    True,
                    alpha,
                    beta
                )

                # Undo move
                board[i] = " "

                # Keep lowest score
                best_score = min(best_score, score)

                # Update beta
                beta = min(beta, best_score)

                # ==================================
                # ALPHA-BETA PRUNING
                # ==================================

                if beta <= alpha:
                    break

        return best_score



# FIND BEST MOVE


def best_move(board):

    # Best score for AI starts at negative infinity
    best_score = -float("inf")

    # No move selected yet
    move = None

    # Initial alpha and beta
    alpha = -float("inf")
    beta = float("inf")

    # Try every board position
    for i in range(9):

        # Only consider empty positions
        if board[i] == " ":

            # Temporarily make AI move
            board[i] = "O"

            # Calculate score of this move
            score = minimax(
                board,
                False,
                alpha,
                beta
            )

            # Undo the move
            board[i] = " "

            # Check whether this is the best move
            if score > best_score:

                best_score = score
                move = i

            # Update alpha
            alpha = max(alpha, best_score)

    return move


# ==========================================
# PRINT BOARD
# ==========================================

def print_board(board):

    print()

    print(
        " " + board[0] +
        " | " + board[1] +
        " | " + board[2]
    )

    print("---+---+---")

    print(
        " " + board[3] +
        " | " + board[4] +
        " | " + board[5]
    )

    print("---+---+---")

    print(
        " " + board[6] +
        " | " + board[7] +
        " | " + board[8]
    )

    print()



# HUMAN MOVE


def human_move(board):

    while True:

        try:

            position = int(
                input("Enter your move (1-9): ")
            )

            # Convert 1-9 to 0-8
            position = position - 1

            # Check range
            if position < 0 or position > 8:

                print(
                    "Please enter a number from 1 to 9."
                )

                continue

            # Check whether position is occupied
            if board[position] != " ":

                print(
                    "That position is already occupied."
                )

                continue

            # Make human move
            board[position] = "X"

            break

        except ValueError:

            print(
                "Please enter a valid number."
            )


# MAIN GAME


def main():

    # Create empty board
    board = [
        " ", " ", " ",
        " ", " ", " ",
        " ", " ", " "
    ]

    print("================================")
    print("    TIC-TAC-TOE AI")
    print("    ALPHA-BETA PRUNING")
    print("================================")

    print()

    print("You are X")
    print("AI is O")

    print()

    print("Board positions:")

    print()

    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")

    print()

    # Human plays first
    current_player = "X"

    while True:

        # ======================================
        # HUMAN TURN
        # ======================================

        if current_player == "X":

            print_board(board)

            human_move(board)

            # Check human victory
            if check_winner(board, "X"):

                print_board(board)

                print("Congratulations! You win!")

                break

            # Check draw
            if " " not in board:

                print_board(board)

                print("It's a draw!")

                break

            # Change player
            current_player = "O"


        # ======================================
        # AI TURN
        # ======================================

        else:

            print()
            print("AI is thinking...")

            # Find best move
            move = best_move(board)

            # Make AI move
            board[move] = "O"

            print(
                "AI chose position:",
                move + 1
            )

            # Check AI victory
            if check_winner(board, "O"):

                print_board(board)

                print("AI wins!")

                break

            # Check draw
            if " " not in board:

                print_board(board)

                print("It's a draw!")

                break

            # Change player
            current_player = "X"


# ==========================================
# START PROGRAM
# ==========================================

if __name__ == "__main__":
    main()
```
